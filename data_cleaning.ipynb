# Fixing Year column ( floa64 to Int64 )

```python
df['Year'].fillna(0, inplace=True)
```

```python
df['Year'] = df['Year'].astype(int)
```

# Creating an Engine Size column

```python
df['EngineSize'] = df['Engine'].str.split(' ', expand=True)[2].astype(float)
```

```python
df.rename(columns={'EngineSize': 'EngineSize(L)'},inplace=True)
```

# Fix the Fuel Consumption column

```python
df['FuelConsumption'].fillna(0, inplace=True)
```

```python
df['FuelConsumption'].replace('-', 0, inplace=True)
```

```python
df['FuelConsumption'] = df['FuelConsumption'].str.split(' ', expand=True)[0].astype(float)
```

```python
df.rename(columns={'FuelConsumption': 'FuelConsumption(L/100 KM)'}, inplace=True)
```

## Expand the Location Column

```python
df['State'] = df['Location'].str.split(',', expand=True)[1]
df['Suburb'] = df['Location'].str.split(',', expand=True)[0]
```

Fixing Cylinders in Engine Column

```python
df['CylindersinEngine'] = df['CylindersinEngine'].str.split(' ', expand=True)[0]

df.rename(columns = {'CylindersinEngine' : 'CylindersinEngine(cyl)'})
```

```python
df = df.loc[df['Price'] != 'POA']
```
